# 02 - YOLO ONNX Mask Detection Evaluation
This notebook evaluates a trained Ultralytics YOLO ONNX model (`best_yolo.onnx`) on a mask dataset.

```
# Install if required
# !pip install ultralytics onnxruntime opencv-python matplotlib scikit-learn tqdm
```

```
from ultralytics import YOLO
from pathlib import Path
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
from tqdm import tqdm
import time
import numpy as np
```

```
MODEL_PATH='best_yolo.onnx'
model=YOLO(MODEL_PATH)

print("Model loaded")

try:
    print("Class names:",model.names)
except Exception as e:
    print("Could not read class names:",e)
```

```
DATASET=Path("../datasets/mask/data")

GT={
    "with_mask":0,
    "without_mask":1
}

image_paths=[]
true_labels=[]

for cls in GT:
    for img in (DATASET/cls).glob("*.*"):
        image_paths.append(img)
        true_labels.append(GT[cls])

print("Images:",len(image_paths))
```

```
pred_labels=[]
times=[]

for img in tqdm(image_paths):

    t0=time.time()

    result=model.predict(
        source=str(img),
        verbose=False
    )[0]

    times.append(time.time()-t0)

    if len(result.boxes)==0:
        pred_labels.append(-1)
        continue

    cls=int(result.boxes.cls[0].item())
    pred_labels.append(cls)

print("Average inference (ms):",np.mean(times)*1000)
print("FPS:",1/np.mean(times))
```

```
valid=[i for i,p in enumerate(pred_labels) if p!=-1]

y_true=[true_labels[i] for i in valid]
y_pred=[pred_labels[i] for i in valid]

print(classification_report(y_true,y_pred,zero_division=0))
```

```
cm=confusion_matrix(y_true,y_pred)

disp=ConfusionMatrixDisplay(cm)
disp.plot()

plt.show()
```

```
acc=np.mean(np.array(y_true)==np.array(y_pred))

print("Accuracy:",acc)
```

## Notes

- If `model.names` reports three classes (e.g. `with_mask`, `without_mask`, `mask_weared_incorrect`), update the `GT` mapping and your dataset accordingly.
- If predictions are shifted (e.g. class IDs differ), print `model.names` and adjust `GT` to match the model's class indices.
- If no detections are found for many images, verify that the model was trained as a detector rather than a classifier and that the input images match the expected format.
